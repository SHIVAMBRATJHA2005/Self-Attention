```
import torch
import torch.nn as nn
import math

class SelfAttention(nn.Module):
    def __init__(self, embedded):
        super(SelfAttention, self).__init__()
        self.embedded = embedded

  def forward(self,x):
      self.query=nn.Linear(embedded,embedded)
      self.key=nn.Linear(embedded,embedded)
      self.value=nn.Linear(embedded,embedded)
      Q=self.query(x)
      K=self.key(x)
      V=self.value(x)
      scores=torch.matmul(Q,K.transpose(-2,-1))
      scores=scores/math.sqrt(self.embedded.size(-1))
      attention=torch.softmax(scores,dim=-1)
      out=torch.matmul(attention,V)
      return out


```
